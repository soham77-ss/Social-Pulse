```python
import pandas as pd

df_clean = pd.read_csv('../data/processed/reddit_clean_posts.csv')
print(f'Clean Unique Discussions: {len(df_clean)}')
df_clean[['brand', 'community_or_context', 'title', 'matched_aspects', 'word_count']].head(8)
```

## 1. Multi-Label Product Aspect Coverage (Research Question #1)

```python
aspect_cols = [c for c in df_clean.columns if c.startswith('aspect_') and c != 'aspect_count']
aspect_summary = df_clean.groupby('brand')[aspect_cols].sum().T
aspect_summary.index = [idx.replace('aspect_', '') for idx in aspect_summary.index]
aspect_summary['Total'] = aspect_summary.sum(axis=1)
aspect_summary.sort_values('Total', ascending=False)
```
